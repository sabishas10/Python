 # Loop Control Statements

## Continue Statement

```python
for i in range(1,11,1):
    print(i)
```

```python
for i in range(1,11,1):
    if i==5:
        continue
    print(i)
```

## Break Statement

```python
for i in range(1,11,1):
    print(i)
```

```python
for i in range(1,21,1):
    if i==16:
        break
    print(i)
```
