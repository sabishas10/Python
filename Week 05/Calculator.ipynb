```python
print("*****Calculator*****")

# Menu
print("1: Addition")
print("2: Substraction")
print("3: Multiplication")
print("4: Division")
print("5: Exit")
print("--------------------")

# User Input
num1=int(input("Enter  the first number:"))
num2=int(input("Enter the second number:"))
choice=int(input("Enter the choice number:"))

# Conditions
if choice==1:
    answer=num1+num2
elif choice==2:
    answer=num1-num2
elif choice==3:
    answer=num1*num2
elif choice==4:
    answer=num1/num2
elif choice==5:
    answer="Goodbye!"
else:
    answer="Invalid Input"
print(answer)
```
